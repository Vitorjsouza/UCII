```python
df = pd.read_excel(
    "vendas_narastore.xlsx", usecols=["Vendedor","Total (R$)"])  # usecols serve como um filtro para colunas, usando o nome ou a posição da coluna
df
```

```python
total = df["Total (R$)"].sum()
media = df["Total (R$)"].mean()
maximo = df["Total (R$)"].max()
minimo = df["Total (R$)"].min()
```

```python
print(f'Total {total}')
print(f'Média {media}')
print(f'Máximo {maximo}')
print(f'Mínimo {minimo}')
```
