```python
df_furto_cel_dp = df.groupby('cisp')['furto_celular'].sum().reset_index() # informação qualitativa utilizo parenteses (''); informação quantitativa utilizo colchetes ['']; reset_index() após o sum() visualizo o index
df_furto_cel_dp = df_furto_cel_dp.sort_values(by='furto_celular', ascending = False) # ascending = False visualiza a linha do menor para o maior, ordeno as linhas da coluna
df_furto_cel_dp
```

```python
df_ap_menores = df.groupby('cisp')['aaapai'].sum().reset_index()
df_ap_menores = df_ap_menores.sort_values(by='aaapai', ascending = False)
df_ap_menores
```

```python
df_rb_veiculos_dp = df.groupby('cisp')['roubo_veiculo'].sum().reset_index()
df_rb_veiculos_dp = df_rb_veiculos_dp.sort_values(by = 'roubo_veiculo', ascending = False)
df_rb_veiculos_dp
```

```python
df_capital = df[df['munic'] == 'Rio de Janeiro']
df_capital = df.loc[df['munic'] == 'Rio de Janeiro'] # Usando .Loc como filtro
df_capital = df.query("munic == 'Rio de Janeiro'") # Usando .Query como filtro melhor de desempenho
df_capital
```

```python
df_bel_rox = df.loc[df['munic'] == 'Belford Roxo']
df_bel_rox = df_bel_rox.groupby('cisp')['roubo_veiculo'].sum().reset_index()
df_bel_rox = df_bel_rox.sort_values(by = 'roubo_veiculo', ascending = False)
df_bel_rox
```

```python
munic_pedido = ['Belford Roxo', 'Nova Iguaçu', 'Queimados']

df_bel_nov_quei = df.loc[df['munic'].isin(munic_pedido)]  # Isin é feito para fazer um filtro usando uma lista como base

df_bel_nov_quei = df_bel_nov_quei.groupby('cisp')['roubo_veiculo'].sum().reset_index()
df_bel_nov_quei = df_bel_nov_quei.sort_values(by = 'roubo_veiculo', ascending = False)
df_bel_nov_quei
```

```python
array_furto_cel_dp = np.array(df_furto_cel_dp['furto_celular'])  # np = numpay # array é para os cálculos do np
array_furto_cel_dp
```

```python
media = np.mean(array_furto_cel_dp) # Código para descobrir a média
mediana = np.median(array_furto_cel_dp) # código para descobrir a mediana

print(f'Média {media}\nMediana {mediana}')
```

```python
minimo = np.min(array_furto_cel_dp)
maximo = np.max(array_furto_cel_dp)
amplitude = maximo - minimo
media = np.mean(array_furto_cel_dp) 
mediana = np.median(array_furto_cel_dp) 
q1 = np.percentile(array_furto_cel_dp, 25)  # Quartis são q1 = 25% ; q2 = 50% ; q3 = 75% 
q2 = np.percentile(array_furto_cel_dp, 50)  # Os Quartis é uma linha de corte
q3 = np.percentile(array_furto_cel_dp, 75)
distancia = (media - mediana) / mediana
iqr = q3 - q1
limite_superior = q3 + (1.5 * iqr)
limite_inferior = q1 - (1.5 * iqr)

print(f'Média {media}')
print(f'Mediana {mediana}')
print(f'Q1: {q1}')
print(f'Q2: {q2}')
print(f'Q3: {q3}')
print(f'{distancia:.2%}')
print(f'Limite Superior {limite_superior}')
print(f'Limite Inferior {limite_inferior}') # Se o limite inferior for negativo, não existem outiliers
```
