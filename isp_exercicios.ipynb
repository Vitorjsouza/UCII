```python
import pandas as pd

import numpy as np
```

```python
url = 'https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv'
df = pd.read_csv(url, encoding = 'latin1', sep=';')
df.columns
```

```python
df_roubo_veiculo_dp = df.groupby('cisp')['roubo_veiculo'].sum().reset_index() # informação qualitativa utilizo parenteses (''); informação quantitativa utilizo colchetes ['']; reset_index() após o sum() visualizo o index
df_roubo_veiculo_dp = df_roubo_veiculo_dp.sort_values(by='roubo_veiculo', ascending = False) # ascending = False visualiza a linha do menor para o maior, ordeno as linhas da coluna
df_roubo_veiculo_dp
```

```python
array_roubo_veiculo_dp = np.array(df_roubo_veiculo_dp['roubo_veiculo'])
```

```python
media = np.mean(array_roubo_veiculo_dp)
mediana = np.median(array_roubo_veiculo_dp)
q1 = np.percentile(array_roubo_veiculo_dp, 25)
q2 = np.percentile(array_roubo_veiculo_dp, 50)
q3 = np.percentile(array_roubo_veiculo_dp, 75)
print(f'Média {media}')
print(f'Mediana {mediana}')
print(f'Q1: {q1}')
print(f'Q2: {q2}')
print(f'Q3: {q3}')
```

```python
distancia = (media - mediana) / mediana  # Número percentual por isso sai um percentual, depois é só formatar para %
print(distancia)

print(f'{distancia:.2%}')
```

```python
df_top_25_roubo = df_roubo_veiculo_dp.loc[df_roubo_veiculo_dp['roubo_veiculo'] >q3]
df_top_25_roubo
```
