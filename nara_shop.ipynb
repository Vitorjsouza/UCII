```python
import pandas as pd

import numpy as np
```

```python
df = pd.read_csv('pedidos_narashop.csv', sep=';' )
df.head()
```

```python
df['valor_pedido'] = df['quantidade'] * df['valor_unitario']  # 1
df
```

```python
df_faturamento_por_categoria = df.groupby("categoria")["valor_total"].sum().reset_index()
df_faturamento_por_categoria = df_faturamento_por_categoria.sort_values(by = "valor_pedido", ascending = False)  # 2
df_faturamento_por_categoria
```

```python
pedidos_auditoria = df[(df["status_pedido"] == "Entregue") & (df["valor_pedido"] > 500)] # 3
pedidos_auditoria
```

```python
df_produto_valor_vendido = df.groupby('produto')['valor_pedido'].sum().reset_index().sort_values(by = 'valor_pedido', ascending=False)  #4
df_produto_valor_vendido
```
