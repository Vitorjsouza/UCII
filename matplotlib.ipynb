```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python

dados = {'categorias': ['Eletronicos', 'Roupas', 'Alimentos', 'Casa'],
         'valores': [4200, 2800, 3600, 1900]}

df = pd.DataFrame(dados)
df
```

```python
plt.bar(df['categorias'], df['valores'])
plt.title('Total de Vendas por categoria - NaraStore')
plt.show
```

```python
df = pd.read_csv('vendas_narastore.csv')
df
```

```python
df_cat_total_vendido = df.groupby('categoria')['total_venda'].sum().reset_index().sort_values(by='total_venda', ascending= False)
df_cat_total_vendido
```

```python
plt.figure(figsize=(10,5)) # sempre antes de definir o gráfico, altere o tamanho da figura
plt.bar(df_cat_total_vendido['categoria'], df_cat_total_vendido['total_venda'])  # qualitativa x quantitaiva
plt.xticks(rotation=45) # grau de angulação do eixo X
plt.show()
```

```python
plt.barh(df_cat_total_vendido['categoria'], df_cat_total_vendido['total_venda'])
plt.title('Total de Vendas por Categoria - NaraStore')
plt.show()
```
