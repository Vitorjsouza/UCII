```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
df = pd.DataFrame({
 'investimento_marketing': [10, 20, 30, 40, 50, 60],
 'vendas': [100, 130, 160, 180, 210, 240]})
X = df[['investimento_marketing']]
y = df['vendas']
```

```python
# Treinar o modelo
modelo = LinearRegression()
modelo.fit(X, y)
# Prever até R$ 80 mil
X_ext = pd.DataFrame({
'investimento_marketing': [80,100,120,150]})
y_pred = modelo.predict(X_ext)
# Gráfico
plt.scatter(df['investimento_marketing'], df['vendas'],
color='green', label='Reais')
plt.plot(X_ext, y_pred,
color='orange', label='Regressão')
plt.show()
```
